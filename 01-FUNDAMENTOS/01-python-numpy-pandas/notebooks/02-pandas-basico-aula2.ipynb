Imagine o cenário: você trabalha no time de dados do Airbnb. O diretor regional te envia uma base de dados crua e diz: "Temos anúncios com valores nulos causados por falhas no app, registros duplicados de proprietários ansiosos e precisamos isolar apenas as acomodações que custam mais de R$ 300 E que aceitam mais de 4 hóspedes para uma campanha premium."

Se você falar em usar loops *for* ou condicionais nativos do Python para varrer essa tabela, a entrevista acaba ali. O mercado exige performance e elegância com Filtros Booleanos e Métodos de Limpeza Nativa

1. Filtros Complexos (Múltiplas Condições)
No Pandas, para combinar filtros, você não usa as palavras *and* ou *or*. Você usa obrigatoriamente os operadores bitwise:


*&* para *E* (todas as condições verdadeiras).


*|* para *OU* (pelo menos uma condição verdadeira).

**Regra de Ouro do Mercado:** Cada condição deve estar, obrigatoriamente, isolada entre parênteses (). Sem eles, o Python se perde na precedência de operadores e estoura um erro.

```python
# Sintaxe Profissional
import pandas as pd
import numpy as np

df = pd.read_csv('dados.csv')
df_premium = df[(df['Preco'] > 300) & (df['Hospedes'] > 4)]
print(df_premium)
```

2. Faxina de Dados (Tratamento de Nulos e Duplicados)
Dados do mundo real são sujos. Em uma sabatina técnica, termos como "Data Wrangling" ou "Data Cleaning" vão aparecer, e você precisa conhecer essas duas funções de cor:


*.drop_duplicates():* Remove linhas idênticas, deixando seu dataset único.


*.fillna(valor):* Substitui todos os valores ausentes (NaN) por um valor padrão (ex: 0 ou a média da coluna) para que seus cálculos futuros não quebrem.

## ⚠️ Erros Clássicos de Iniciantes

1. Esquecer os parênteses nos filtros: Fazer *df[df['Idade'] > 20 & df['Cidade'] == 'SP']*. Vai quebrar.

2. Achar que o Pandas altera o dado original sozinho: Funções como .fillna() ou .drop_duplicates() retornam uma cópia modificada. Para salvar a alteração na mesma variável, você deve reatribuir: df = df.fillna(0) ou usar o argumento inplace=True.

## 🏋️ Exercício Prático 2
Chegou a sua hora de resolver o problema do Airbnb. Você recebeu o seguinte DataFrame de anúncios:

```python
import pandas as pd
import numpy as np

df_airbnb = pd.DataFrame({
    'Quarto': ['Ap Centro', 'Casa Praia', 'Ap Centro', 'Studio Moderno', 'Loft'],
    'Preco': [150, np.nan, 150, 400, 350],
    'Avaliacao': [4.5, 4.8, 4.5, np.nan, 4.9]
})
```

Sua Missão:

1. Escreva o código para eliminar a linha duplicada (o segundo 'Ap Centro').

2. Escreva o código para preencher todos os valores nulos (NaN) com o número 0.

3. Escreva um filtro que retorne apenas os quartos com Preco maior que 200 OU Avaliacao maior que 4.7.

```python
import pandas as pd
import numpy as np

df_airbnb = pd.DataFrame({
    'Quarto': ['Ap Centro', 'Casa Praia', 'Ap Centro', 'Studio Moderno', 'Loft'],
    'Preco': [150, np.nan, 150, 400, 350],
    'Avaliacao': [4.5, 4.8, 4.5, np.nan, 4.9]
})

# Missão 1 - Identificar os quartos duplicados e eliminar as linhas duplicadas.

# df_airbnb = df_airbnb[~df_airbnb['Quarto'].duplicated()]
df_airbnb = df_airbnb.drop_duplicates()

# Missão 2 - Prencher todos os valores nulos (NaN) com 0.

df_airbnb.fillna(0, inplace=True)

# Missão 3 - Fazer um filtro que retorne apens os quartos com preço maior que 200 ou avaliação maior que 4.7.

df_airbnb_filtrado = df_airbnb[(df_airbnb['Preco'] > 200) | (df_airbnb['Avaliacao'] > 4.7)]
display(df_airbnb_filtrado)
```

A sua lógica com ~df_airbnb['Quarto'].duplicated() funciona matematicamente para limpar duplicatas olhando apenas para a coluna 'Quarto', mas ela foge do padrão de mercado para limpeza de linhas idênticas inteiras. Em uma sabatina técnica, termos como "Data Wrangling" ou "Data Cleaning" vão aparecer, e você precisa conhecer essas funções nativas de cor. O comando exato e otimizado que o mercado espera é simplesmente a aplicação do método direto na tabela:

```python
# A forma esperada na entrevista técnica
df_airbnb = df_airbnb.drop_duplicates()
```

## 📝 Considerações Finais do Bloco 2 (Folha de Consulta Rápida)

Antes de mudarmos de assunto, guarde este resumo mental, pois ele representa a maior parte do seu trabalho de limpeza:

1. Filtros Condicionais: Nunca use and/or. Use sempre & e |.

2. A Regra de Ouro: Isole cada condição obrigatoriamente entre parênteses () para o Python não se perder na matemática.

3. Tratamento de Nulos: .fillna(valor) tapa os buracos (os NaN) do seu dataset para não quebrar cálculos.

4. Tratamento de Duplicatas: .drop_duplicates() limpa registros que foram inseridos mais de uma vez no sistema.

5. Salvando Alterações: Lembre-se que esses métodos de limpeza retornam uma cópia. Sempre reatribua à variável (df = df...) ou use inplace=True.
