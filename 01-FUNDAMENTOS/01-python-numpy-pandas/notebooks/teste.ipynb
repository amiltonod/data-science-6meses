```python
import pandas as pd

df_pedidos = pd.DataFrame({'ID_Restaurante': [1, 2, 1, 3, 2], 'Valor_Pedido': [50, 120, 80, 45, 200]})
df_restaurantes = pd.DataFrame({'ID_Restaurante': [1, 2, 3], 'Nome_Restaurante': ['Pizzaria X', 'Sushi Y', 'Hamburgueria Z']})

df_completo = pd.merge(df_pedidos, df_restaurantes, on='ID_Restaurante', how='left')
faturamento = df_completo.groupby('Nome_Restaurante')['Valor_Pedido'].sum()

print(faturamento)
```
