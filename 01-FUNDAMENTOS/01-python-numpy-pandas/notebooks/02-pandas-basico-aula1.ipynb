# 🚀 Aula 1: DataFrames, Series e Seleção Cirúrgica
Imagine que você acabou de entrar na Netflix. O seu gestor te entrega um arquivo CSV com 50.000 linhas contendo o histórico de acessos dos usuários e diz: "Preciso que você isole os dados do usuário ID 1042 e descubra quais colunas temos disponíveis."

No Pandas, você lidará com duas estruturas fundamentais:

Series: É uma única coluna. Pense nela como uma lista do Python, mas com um índice (rótulo) para cada linha.

DataFrame: É a tabela inteira. Um conjunto de várias Series compartilhando o mesmo índice.

Como acessar os dados (O divisor de águas nas entrevistas)
Os iniciantes se batem muito para extrair dados específicos. O mercado exige que você domine a diferença entre **.loc** e **.iloc:**

.loc[linha, coluna]: Busca pelo NOME do rótulo.

.iloc[linha, coluna]: Busca pela POSIÇÃO numérica (Índice Inteiro), exatamente como fatiamento de listas.

```python
import pandas as pd

# Simulação de uma tabela de RH
df = pd.DataFrame({
    'Nome': ['Alice', 'Bob', 'Carol'],
    'Idade': [25, 30, 28],
    'Cidade': ['SP', 'RJ', 'MG']
}, index=['func_101', 'func_102', 'func_103'])

# Seleção por Nome do Índice (.loc)
print(df.loc['func_101', 'Nome'])  # Retorna: Alice

# Seleção por Posição Física (.iloc)
print(df.iloc[0, 0])  # Retorna: Alice (Linha 0, Coluna 0)
```

## ⚠️ Erro Clássico de Iniciante

Tentar modificar um pedaço do DataFrame fazendo *df[df['Idade'] > 25]['Cidade'] = 'Desconhecido'.*
Isso gera o famoso erro/aviso: SettingWithCopyWarning.
O Pandas se perde se você está modificando uma cópia ou o dado original.
A regra de ouro do mercado é: para alterar dados selecionados, use sempre .loc.

Escreva mentalmente ou no seu ambiente de código a solução para o seguinte caso de uso da Uber:

Você tem o DataFrame df_corridas abaixo:

```python
df_corridas = pd.DataFrame({
    'Motorista': ['Carlos', 'Sonia', 'Pedro'],
    'Faturamento': [1200, 1500, 900]
}, index=['id_0', 'id_1', 'id_2'])
```

### Sua missão:

>1 - Escreva a linha de comando usando .loc para retornar apenas o faturamento da Sonia (utilizando o índice dela).

>2 - Escreva a linha de comando usando .iloc para extrair os dados da primeira linha inteira (Carlos).

```python
import pandas as pd

# Simulação de uma tabela de corridas
df_corridas = pd.DataFrame({
    'Motorista': ['Carlos', 'Sonia', 'Pedro'],
    'Faturamento': [1200, 1500, 900]
}, index=['id_0', 'id_1', 'id_2'])

# Seleção por Nome do Índice (.loc)
print(df_corridas.loc['id_1', 'Faturamento'])  # Retorna: 1500

# Seleção da primeira lina inteira
print(df_corridas.iloc[0:1])  # Retorna a primeira linha inteira do DataFrame
```
