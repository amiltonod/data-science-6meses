```python
import numpy as np

# 1. Transformando uma lista de faturamento em Array
faturamento_lista = [1020.50, 3040.00, 1500.25, 5000.00]
faturamento_array = np.array(faturamento_lista)

print(faturamento_array)
print(faturamento_array.dtype)  # Mostra o tipo de dado: ex float64
```

```python
# 🏋️ Exercício Prático 1

#Você recebeu a tarefa de criar uma simulação para o estoque de uma loja.
#1 - Crie um array NumPy chamado IDs_produtos que vá de 100 até 150 (inclusive), pulando de 2 em 2.
#2 - Crie um array chamado estoque_inicial preenchido com o número 10, contendo 26 elementos (o mesmo tamanho do array de IDs). 

IDs_produtos = np.arange(100, 152, 2)

estoque_inicial = np.full(26, 10)
# ou estoque_inicial = np.ones(26) * 10


print("IDs:", IDs_produtos)
print("Qtd de IDs:", len(IDs_produtos))
print("Estoque:", estoque_inicial)
```
