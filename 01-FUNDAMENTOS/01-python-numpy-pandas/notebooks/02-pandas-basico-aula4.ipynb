No dia a dia, raramente vais criar DataFrames à mão digitando dicionários Python. Os dados das empresas estão guardados em ficheiros CSV (vindos de sistemas ou bases de dados) ou em folhas de cálculo Excel (geradas pelas equipas de negócio).

O Pandas foi desenhado para engolir estes ficheiros com apenas uma linha de código. Vamos focar-nos nos 20% de conhecimento que resolvem 80% dos problemas de importação (e que os recrutadores adoram testar).

1. Ler Ficheiros CSV (pd.read_csv)
O CSV (Comma-Separated Values) é um ficheiro de texto simples onde os dados são separados por um caractere.

O Comando Básico:

```python
df = pd.read_csv('vendas.csv')
```

## ⚠️ Os Dois Erros Clássicos que Eliminam Candidatos em Testes Técnicos:

1. O Separador Errado (sep): Por padrão, o Pandas assume que o separador é uma vírgula (,). 
No entanto, em países de língua portuguesa (e muitos na Europa), o Excel usa o ponto e vírgula (;) para não confundir com a vírgula dos números decimais (ex: 10,50). 
Se o teu ficheiro vier com ; e não avisares o Pandas, ele vai ler a linha inteira como se fosse uma única coluna de texto e o teu código vai quebrar.

A Solução: *df = pd.read_csv('vendas.csv', sep=';')*

2. O Erro de Codificação (encoding):
Se o teu ficheiro contiver palavras com acentos ou cedilhas (como "Promoção", "Preço", "Sábado"), o Pandas pode dar um erro gigante de UnicodeDecodeError.
Isto acontece porque o Python tenta ler o ficheiro num formato padrão (geralmente UTF-8) mas o sistema da empresa exportou noutro.

A Solução: Forçar o encoding do Windows/Excel usando latin1 ou iso-8859-1.

O Código de Elite: *df = pd.read_csv('vendas.csv', sep=';', encoding='latin1')*

2. Ler Ficheiros Excel *(pd.read_excel)*
O Excel (.xlsx) é um formato mais complexo. O Pandas lê-o nativamente, mas precisas de ter atenção a um detalhe: um ficheiro Excel pode ter várias abas (folhas/sheets) lá dentro.

O Comando Básico:

```python
df = pd.read_excel('relatorio_financeiro.xlsx')
```
