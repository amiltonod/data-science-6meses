### 1. O Poder do GroupBy (A Máquina de Relatórios)
O .groupby() é baseado no conceito de Split-Apply-Combine (Separar, Aplicar, Combinar).

Você separa a tabela baseada em uma coluna (ex: Região), aplica uma função matemática (ex: Média) em outra coluna (ex: Vendas), e combina tudo em um novo resumo.

A Sintaxe de Ouro do Mercado:

```python
# df.groupby('Coluna_Agrupadora')['Coluna_Numerica'].funcao_matematica()

# Exemplo: Média de idade por cidade
resumo = df.groupby('Cidade')['Idade'].mean()
```

Funções de agregação (.agg) mais usadas: .sum() (Soma), .mean() (Média), .count() (Contagem), .max() (Máximo).

### 2. Cruzamento de Tabelas (.merge)
Raramente todos os dados de uma empresa estão em um único arquivo. O time de Vendas tem uma tabela com o ID_Cliente e o time de Cadastro tem outra tabela com o Nome e a Cidade desse mesmo ID_Cliente. Você precisa cruzar essas tabelas. No SQL isso se chama JOIN, no Pandas chamamos de merge

```python
# Cruzando a tabela de Vendas com a tabela de Clientes usando o ID como chave
df_completo = pd.merge(df_vendas, df_clientes, on='ID_Cliente', how='left')
```

O argumento how='left' (ou inner, right) dita como os dados se comportarão caso um ID exista em uma tabela e não na outra. No dia a dia, o left (manter todos da tabela da esquerda) é o mais utilizado.

## ⚠️ Erro Clássico de Iniciante
Fazer um .groupby() e esquecer de selecionar a coluna numérica depois dele. Fazer df.groupby('Cidade').mean() vai tentar calcular a média de todas as colunas da tabela, o que pode quebrar o código ou gerar um processamento lento se existirem colunas de texto no meio. O iFood te enviou duas tabelas simplificadas referentes ao fechamento da noite de ontem.

Tabela de Pedidos:

```python
import pandas as pd

df_pedidos = pd.DataFrame({
    'ID_Restaurante': [1, 2, 1, 3, 2],
    'Valor_Pedido': [50, 120, 80, 45, 200]
})

#

df_restaurantes = pd.DataFrame({
    'ID_Restaurante': [1, 2, 3],
    'Nome_Restaurante': ['Pizzaria X', 'Sushi Y', 'Hamburgueria Z']
})
```

Sua Missão Final:

1. Escreva o código para fazer um Merge entre df_pedidos e df_restaurantes para que saibamos o nome do restaurante em cada pedido.

2. Usando o DataFrame cruzado resultante do passo 1, escreva o código com .groupby() para calcular o Soma Total (sum) do Valor_Pedido para cada Nome_Restaurante.

```python

import pandas as pd
import numpy as np

df_pedidos = pd.DataFrame({
'ID_Restaurante': [1, 2, 1, 3, 2],
'Valor_Pedido': [50, 120, 80, 45, 200]
})

# Tabela de Restaurantes:

df_restaurantes = pd.DataFrame({
'ID_Restaurante': [1, 2, 3],
'Nome_Restaurante': ['Pizzaria X', 'Sushi Y', 'Hamburgueria Z']
})

# Missão 1: Fazer um merge entre df_pedidos e df_restautantes para que saibamos o nome do restaurante de cada pedido
df_completo = pd.merge(df_pedidos, df_restaurantes, on='ID_Restaurante', how='left')
display(df_completo)

# Missão 2: Calculaar a soma total do valor do pedido de cada restaurante
faturamento = df_completo.groupby('Nome_Restaurante')['Valor_Pedido'].sum()
display(faturamento)

```
