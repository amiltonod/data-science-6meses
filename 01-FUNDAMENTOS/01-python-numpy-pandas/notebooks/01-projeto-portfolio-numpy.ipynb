Você recebeu uma matriz que representa os dados comerciais de 5 filiais de uma empresa no último trimestre. Cada linha é uma filial, e as colunas são: *[ID da Filial, Qtd de Clientes Cancelados (Churn), Faturamento Total (R$)].*

## 📋 Suas 3 Missões no Jupyter:

Fatiamento Profissional: Isole a coluna de Churn (coluna de índice 1) em uma variável chamada churn e a coluna de Faturamento (coluna de índice 2) em uma variável chamada faturamento.

Filtro de Risco: Use indexação booleana para descobrir os IDs das filiais que estão com o Churn crítico, ou seja, maior que 40.

Métrica de Diretoria: Calcule a média de faturamento dessas 5 filiais usando a função correta do NumPy.

### 🎯 Mini Simulado de Entrevista Técnica

Finja que eu sou o entrevistador da vaga dos seus sonhos. Olhando para esses dados, te faço duas perguntas rápidas para testar sua mentalidade de alto nível:

```python
import numpy as np

# --- CONFIGURAÇÃO INICIAL ---
# A matriz oficial do projeto com os dados brutificados das filiais
dados_comerciais = np.array([
    [10, 45, 150000],  # Filial ID 10
    [11, 12, 320000],  # Filial ID 11
    [12, 85, 95000],   # Filial ID 12
    [13, 30, 410000],  # Filial ID 13
    [14, 60, 210000]   # Filial ID 14
])


# 📋 MISSÃO 1: Fatiamento Profissional
# Isolamos a coluna de índice 1 (Churn) e a coluna de índice 2 (Faturamento) de TODAS as linhas (:)
churn = dados_comerciais[:, 1]
faturamento = dados_comerciais[:, 2]

print("--- REQUISITO 1 ---")
print("Variável 'churn' isolada:", churn)
print("Variável 'faturamento' isolada:", faturamento)
print("-" * 30)


# 📋 MISSÃO 2: Filtro de Risco (Indexação Booleana)
# Passo A: Criamos uma máscara que verifica quem tem churn maior que 40
mascara_critica = churn > 40

# Passo B: Aplicamos a máscara na matriz para extrair apenas os IDs (Coluna de índice 0)
ids_filiais_criticas = dados_comerciais[mascara_critica, 0]

print("--- REQUISITO 2 ---")
print("IDs das filiais com Churn crítico (> 40):", ids_filiais_criticas)
print("-" * 30)


# 📋 MISSÃO 3: Métrica de Diretoria
# Calculamos a média usando a variável 'faturamento' que isolamos no passo 1
media_faturamento = np.mean(faturamento)

print("--- REQUISITO 3 ---")
print(f"Média de faturamento das 5 filiais: R$ {media_faturamento:.2f}")
print("-" * 30)
```

# 📝 Resumo Final do Treinamento de Elite (Sua Folha de Consulta)

*np.array(): Cria seu array/matriz.*

*[:, 2]: Fatia a coluna de índice 2 de todas as linhas.*

*(array > 40): Cria uma máscara booleana (True/False).*

*array[mascara]: Filtra o array usando a máscara.*

*np.sum(), np.mean(): Soma e Média.*
