#Python

## 2. O Conceito de Eixos (axis) em Tabelas (Matrizes)
Isso aqui separa os amadores dos profissionais. Se você tem uma matriz 2D (linhas e colunas), como você calcula a soma apenas das colunas ou apenas das linhas?

axis=0: Operação vertical (calcula esmagando as colunas).

axis=1: Operação horizontal (calcula esmagando as linhas).


```python
import numpy as np

# Python
# Linhas = Lojas diferentes | Colunas = Vendas de Jan, Fev, Mar
vendas_lojas = np.array([
    [10, 15, 20],  # Loja A
    [30, 5,  12]   # Loja B
])

# Qual o total vendido por MÊS? (Soma vertical nas colunas)
soma_meses = np.sum(vendas_lojas, axis=0)
print(soma_meses) # Saída: [40 20 32] -> (Jan: 40, Fev: 20, Mar: 32)
```

### ⚠️ Erro Clássico de Iniciante
Confundir axis=0 com axis=1. Lembre-se do macete de memorização para entrevistas: axis=0 lembra a letra 'O' de "Olhar para baixo" (vertical).

# 🏋️ Exercício Prático 3 (Para fechar o conteúdo técnico)
Uma startup de entregas registrou os seguintes tempos de entrega (em minutos) de 3 motoboys em 4 entregas diferentes cada um. Isso gerou uma matriz 2D (3 linhas, 4 colunas):

Python
Linhas = Motoboys (Motoboy 0, Motoboy 1, Motoboy 2)
Colunas = Tempo de cada uma das 4 entregas realizadas
tempos_entregas = np.array([
    [15, 22, 18, 30],  # Motoboy 0
    [12, 14, 15, 11],  # Motoboy 1
    [25, 28, 32, 20]   # Motoboy 2
])
O Diretor de Operações precisa de duas métricas rápidas:

1 -Qual é a média de tempo de entrega de cada um dos 3 motoboys? (Dica: o resultado deve ser um array com 3 valores, um para cada linha).

2 -Qual foi o tempo da entrega mais rápida (mínima) registrada na empresa inteira?

Quais funções do NumPy e qual configuração de axis você usaria para responder ao Diretor?

---

### ---> Primeira tentativa de lógica.
print(f'Média motoboy 0: {np.mean(tempos_entregas[:, 0]):.2f}')

print(f'Média motoboy 1: {np.mean(tempos_entregas[:, 1]):.2f}')

print(f'Média motoboy 2: {np.mean(tempos_entregas[:, 2]):.2f}')

print(f'Tempo de entrega mais rápido: {np.min(tempos_entregas):.2f}')


### ---> Refatoração: A lógica para o tempo mais rápido (np.min) está perfeita e 100% correta! Você extraiu o menor valor de toda a matriz com maestria.

No entanto, no cálculo das médias dos motoboys, você caiu em uma pegadinha clássica de fatiamento de matrizes que elimina muitos candidatos em testes técnicos. Vamos analisar o que aconteceu:

Seu código usou [:, 0], [:, 1] e [:, 2]. No NumPy, o primeiro argumento : significa "todas as linhas" e o segundo número indica a coluna.
Portanto, o seu código calculou a média das colunas (as primeiras 3 entregas), misturando os tempos de motoboys diferentes!

Lembrando da estrutura da nossa matriz:

Linhas (0, 1, 2) = Os Motoboys.

Colunas (0, 1, 2, 3) = As Entregas.



```python
import numpy as np

tempos_entregas = np.array([
    [15, 22, 18, 30],  # Motoboy 0
    [12, 14, 15, 11],  # Motoboy 1
    [25, 28, 32, 20]   # Motoboy 2
])

# 1. Média de cada motoboy de uma vez só (axis=1 opera nas LINHAS)

medias_motoboys = np.mean(tempos_entregas, axis=1)

print(f"Média do Motoboy 0: {medias_motoboys[0]:.2f}")
print(f"Média do Motoboy 1: {medias_motoboys[1]:.2f}")
print(f"Média do Motoboy 2: {medias_motoboys[2]:.2f}")

# 2. Tempo mais rápido (Seu acerto!)
print(f"Tempo de entrega mais rápido: {np.min(tempos_entregas):.2f}")
```
