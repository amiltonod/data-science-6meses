```python
import numpy as np

# Colunas: [ID Filial, Vendas (unidades), Faturamento (R$), Devoluções]
farmacias = np.array([
    [1, 320, 48000, 15],
    [2, 510, 76500,  8],
    [3, 190, 28500, 42],
    [4, 430, 64500, 20]
])

# Missão 1: Fatiamento, isolar as colunas de faturamento em váriaveis separadas.

faturamento = farmacias[:, 2]
devolucoes = farmacias[:, 3]
print('--------Requisito 1--------')
print(faturamento)
print(devolucoes)
print('---------------------------')

# Missão 2: Filtro de risco, identificação dos IDs das filiais com devoluções acima de 18

# Passo A: Criamos uma máscara que verifica quem tem devolucoes maior que 18
mascara_critica = devolucoes > 18

# Passo B: Aplicamos a máscara na matriz para extrair apenas os IDs (Coluna de índice 0)
ids_filiais_criticas = farmacias[mascara_critica, 0]

print("--------Requisito 2--------")
print("IDs das filiais com devoluções críticas (> 18) que serão auditadas:", ids_filiais_criticas)
print("---------------------------")

# Missão 3: Calcular média total de faturamento total da da rede e média por unidade

# Calculando faturamento total
total_faturamento = np.sum(faturamento)

# Calculando média das unidades
media_unidades = np.mean(faturamento)

print('--------Requisito 3--------')
print('Faturamento total da rede: R$', total_faturamento)
print('Media total das filiaos: R$', media_unidades)
print('---------------------------')


# Missão 4: Descobrir qual indice de maior faturamento de filial e exibir os dados desse faturamento.

maior_faturamento = np.argmax(faturamento)

print('--------Requisito 4--------')
print('Indice do maior faturamento', maior_faturamento)
print('Maior faturamento: R$', farmacias[maior_faturamento, :])
print('---------------------------')
```
